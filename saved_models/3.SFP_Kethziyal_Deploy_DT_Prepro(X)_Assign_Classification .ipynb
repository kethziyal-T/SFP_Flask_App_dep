```python
import pandas as pd
import numpy as np
from imblearn.over_sampling import SMOTE
```

```python
dataset=pd.read_csv('SensorFault_Detector.csv',nrows=20000)
dataset=dataset.drop(['Machine_ID','Laser_Intensity','Hydraulic_Pressure_bar','Coolant_Flow_L_min','Heat_Index'],axis=1)
dataset=pd.get_dummies(dataset,drop_first=True,dtype=int)
dataset['AI_Supervision']=dataset['AI_Supervision'].astype('int64')
dataset['Failure_Within_7_Days']=dataset['Failure_Within_7_Days'].astype('int64')
prep=dataset.to_csv('sensor_failure_analysis.csv', index=False)
```

```python
indep_X=dataset[[ 'Operational_Hours', 'Temperature_C', 'Vibration_mms', 'Oil_Level_pct', 'Remaining_Useful_Life_days']]
indep_X
```

```python
dep_Y=dataset[['Failure_Within_7_Days']]
dep_Y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(indep_X, dep_Y,test_size=0.25,random_state=42)
```

```python
dep_Y = np.ravel(dep_Y)
    
# Balance training split only
smote = SMOTE(random_state=42)
X_train_res, y_train_res = smote.fit_resample(X_train, y_train)

# Normalize features cleanly
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train_scaled = sc.fit_transform(X_train_res)
X_test_scaled = sc.transform(X_test)
```
